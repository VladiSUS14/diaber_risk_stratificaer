```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "diabetes_data.xlsx"

print("Текущая папка:", Path.cwd())
print("Корень проекта:", PROJECT_ROOT)
print("Путь к файлу:", DATA_PATH)
print("Файл существует:", DATA_PATH.exists())
```

```python
df = pd.read_excel(DATA_PATH)

print("Размер таблицы:", df.shape)
df.head()
```

```python
for i, col in enumerate(df.columns, start=1):
    print(i, repr(col))
```

```python
df.info()
```

```python
patient_col = "№"

print("Количество строк наблюдений:", len(df))
print("Количество уникальных пациентов:", df[patient_col].nunique())

visits_per_patient = df[patient_col].value_counts().sort_index()

print("\nМинимум визитов на пациента:", visits_per_patient.min())
print("Максимум визитов на пациента:", visits_per_patient.max())
print("Среднее количество визитов:", round(visits_per_patient.mean(), 2))

visits_per_patient.describe()
```

```python
missing = df.isna().sum().reset_index()
missing.columns = ["Столбец", "Количество пропусков"]
missing["Доля пропусков, %"] = (missing["Количество пропусков"] / len(df) * 100).round(2)

missing = missing.sort_values("Доля пропусков, %", ascending=False)

missing
```

```python
special_missing_values = ["—", "-", "НД", "нд", "Нет данных", "нет данных", ""]

for value in special_missing_values:
    count = (df == value).sum().sum()
    print(f"'{value}': {count}")
```

```python
binary_cols = [
    "Пол",
    "Терапия",
    "Кетоацидоз",
    "Ангиопатия",
    "Катаракта",
    "Полинейропатия",
    "Стеатоз",
    "Липодистрофии"
]

for col in binary_cols:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False).sort_index())
```
